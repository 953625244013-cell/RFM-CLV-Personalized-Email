```python
# Cell 1: Test Customers from Different Segments

test_customers = [
    {"name": "Arun", "segment": "Champions"},
    {"name": "Priya", "segment": "Loyal Customers"},
    {"name": "Kavi", "segment": "Potential Loyalists"},
    {"name": "Dharshini", "segment": "At Risk"},
    {"name": "Rahul", "segment": "Lost Customers"},
    {"name": "Nithya", "segment": "Others"}
]

print("Testing email generation for different customer segments:")
print(f"Total customers to test: {len(test_customers)}")

for customer in test_customers:
    print(customer)
```

```python
# Cell 2: Segment-Specific Promotions

promotions = {
    "Champions": "20% exclusive discount",
    "Loyal Customers": "15% loyalty discount",
    "Potential Loyalists": "10% special discount",
    "At Risk": "25% win-back discount",
    "Lost Customers": "30% re-engagement discount",
    "Others": "10% special offer"
}

for segment, offer in promotions.items():
    print(f"{segment}: {offer}")
```

```python
# Cell 3: Create Segment-Specific Prompts

def create_prompt(customer_name, segment, promotion):
    prompt = f"""
Create a personalized promotional email for the customer.

Customer Name: {customer_name}
Customer Segment: {segment}
Promotion: {promotion}

Requirements:
- Use the customer's name.
- Make the email relevant to the customer segment.
- Clearly mention the promotion.
- Keep the email friendly and professional.
- Include a subject line.
- Keep the email concise.

Return the result as:
Subject: <subject>
Email: <email body>
"""
    return prompt


# Test prompt creation
for customer in test_customers:
    name = customer["name"]
    segment = customer["segment"]
    promotion = promotions[segment]

    prompt = create_prompt(name, segment, promotion)

    print("=" * 70)
    print(prompt)
```

```python
# Cell 4: Generate Email Prompts for Testing

generated_emails = []

for customer in test_customers:
    name = customer["name"]
    segment = customer["segment"]
    promotion = promotions[segment]

    prompt = create_prompt(
        customer_name=name,
        segment=segment,
        promotion=promotion
    )

    generated_emails.append({
        "customer": name,
        "segment": segment,
        "promotion": promotion,
        "prompt": prompt
    })

print("Prompts created successfully!")
print("Total prompts:", len(generated_emails))
```

```python
# Cell 5: Display Test Prompts

for item in generated_emails:

    print("=" * 70)
    print("CUSTOMER :", item["customer"])
    print("SEGMENT  :", item["segment"])
    print("PROMOTION:", item["promotion"])
    print("-" * 70)
    print(item["prompt"])
    print()
```

```python
# Cell 6: Verify All Customer Segments

print("CUSTOMER SEGMENT TEST")
print("=" * 70)

for item in generated_emails:
    print(f"Customer  : {item['customer']}")
    print(f"Segment   : {item['segment']}")
    print(f"Promotion : {item['promotion']}")
    print("-" * 70)

print("\nTotal customers tested:", len(generated_emails))
print("Total segments tested:", len(set(item["segment"] for item in generated_emails)))
```

```python
# Cell 7: Generate Segment-Specific Email Templates

def generate_test_email(customer_name, segment, promotion):

    messages = {
        "Champions":
            "Thank you for being one of our most valued customers!",

        "Loyal Customers":
            "We truly appreciate your continued loyalty and support!",

        "Potential Loyalists":
            "We would love to see you shop with us again!",

        "At Risk":
            "We miss you! Come back and enjoy this special offer.",

        "Lost Customers":
            "We would love to welcome you back with a special offer!",

        "Others":
            "Enjoy this special offer on your next purchase!"
    }

    message = messages.get(
        segment,
        "Thank you for being our valued customer!"
    )

    email = f"""
Subject: Special Offer Just for You!

Hi {customer_name},

{message}

As a {segment} customer, you can enjoy
{promotion} on your next purchase.

Don't miss this special opportunity!

Best regards,
Customer Team
"""

    return email


for customer in test_customers:

    name = customer["name"]
    segment = customer["segment"]
    promotion = promotions[segment]

    email = generate_test_email(
        name,
        segment,
        promotion
    )

    print("=" * 70)
    print("CUSTOMER:", name)
    print("SEGMENT :", segment)
    print(email)
```

```python
# Cell 8: Store Generated Emails

import pandas as pd

test_results = []

for customer in test_customers:

    name = customer["name"]
    segment = customer["segment"]
    promotion = promotions[segment]

    email = generate_test_email(
        name,
        segment,
        promotion
    )

    test_results.append({
        "Customer": name,
        "Segment": segment,
        "Promotion": promotion,
        "Generated_Email": email
    })

results_df = pd.DataFrame(test_results)

print("Email generation test completed!")
print("Total emails:", len(results_df))

results_df
```

```python
# Cell 9: Validate Email Generation

print("EMAIL GENERATION VALIDATION")
print("=" * 60)

for index, row in results_df.iterrows():

    if pd.notna(row["Generated_Email"]) and row["Generated_Email"].strip() != "":
        status = "SUCCESS"
    else:
        status = "FAILED"

    print(f"Customer : {row['Customer']}")
    print(f"Segment  : {row['Segment']}")
    print(f"Status   : {status}")
    print("-" * 60)

successful = results_df["Generated_Email"].notna().sum()

print("\nTotal emails generated :", len(results_df))
print("Successful emails      :", successful)
print("Failed emails          :", len(results_df) - successful)
```

```python
# Cell 10: Display Final Personalized Emails

for index, row in results_df.iterrows():

    print("\n" + "=" * 80)
    print(f"CUSTOMER : {row['Customer']}")
    print(f"SEGMENT  : {row['Segment']}")
    print(f"PROMOTION: {row['Promotion']}")
    print("=" * 80)

    print(row["Generated_Email"])
```

```python
# Cell 11: Export Email Test Results

output_file = "customer_segment_email_test_results.csv"

results_df.to_csv(output_file, index=False)

print("Test results saved successfully!")
print("File:", output_file)
```

```python
# Cell 13: Segment-wise Email Generation Summary

segment_summary = (
    results_df
    .groupby("Segment")
    .size()
    .reset_index(name="Email_Count")
)

print("SEGMENT-WISE EMAIL GENERATION SUMMARY")
print("=" * 60)

print(segment_summary.to_string(index=False))

print("\nTotal emails generated:", segment_summary["Email_Count"].sum())
```

```python
# Cell 14: Check Email Personalization

print("PERSONALIZATION CHECK")
print("=" * 70)

for index, row in results_df.iterrows():

    email = row["Generated_Email"]

    name_check = row["Customer"] in email
    segment_check = row["Segment"] in email
    promotion_check = row["Promotion"] in email

    if name_check and segment_check and promotion_check:
        status = "PASS"
    else:
        status = "CHECK"

    print(f"Customer : {row['Customer']}")
    print(f"Segment  : {row['Segment']}")
    print(f"Name     : {'YES' if name_check else 'NO'}")
    print(f"Segment  : {'YES' if segment_check else 'NO'}")
    print(f"Promotion: {'YES' if promotion_check else 'NO'}")
    print(f"Status   : {status}")
    print("-" * 70)
```

```python
# Cell 15: Final Email Generation Test Report

total_customers = len(results_df)

successful_emails = 0

for index, row in results_df.iterrows():
    email = row["Generated_Email"]

    if (
        row["Customer"] in email
        and row["Segment"] in email
        and row["Promotion"] in email
    ):
        successful_emails += 1

failed_emails = total_customers - successful_emails

print("=" * 70)
print("FINAL EMAIL GENERATION TEST REPORT")
print("=" * 70)

print("Total Customers Tested :", total_customers)
print("Successful Emails      :", successful_emails)
print("Failed Emails          :", failed_emails)

if failed_emails == 0:
    print("\nRESULT: ALL EMAIL GENERATION TESTS PASSED")
else:
    print("\nRESULT: SOME EMAIL TESTS NEED REVIEW")
```

```python
# Cell 16: Save Final Test Report

final_report = pd.DataFrame({
    "Metric": [
        "Total Customers Tested",
        "Successful Emails",
        "Failed Emails",
        "Segments Tested"
    ],
    "Value": [
        total_customers,
        successful_emails,
        failed_emails,
        results_df["Segment"].nunique()
    ]
})

final_report.to_csv(
    "email_generation_final_report.csv",
    index=False
)

print("Final test report saved successfully!")
print("\n")
print(final_report)
```
