```python
analysis_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)

print("Analysis Date:", analysis_date)
```

```python
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (analysis_date - x.max()).days,
    'InvoiceNo': 'nunique',
    'TotalAmount': 'sum'
})

rfm.columns = ['Recency', 'Frequency', 'Monetary']

rfm.head()
```

```python
rfm = rfm[rfm['Monetary'] > 0]
rfm = rfm[rfm['Frequency'] > 0]

print("Customers after cleaning:", len(rfm))
rfm.head()
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm.head()
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

print("RFM individual scores created!")
```

```python
rfm['RFM_Score'] = (
    rfm['R_Score'] +
    rfm['F_Score'] +
    rfm['M_Score']
)

print("RFM Score created!")

rfm[
    ['Recency', 'Frequency', 'Monetary',
     'R_Score', 'F_Score', 'M_Score', 'RFM_Score']
].head()
```

```python
def get_segment(score):

    if score >= 13:
        return "Champions"

    elif score >= 11:
        return "Loyal Customers"

    elif score >= 9:
        return "Potential Loyalists"

    elif score >= 7:
        return "At Risk"

    elif score >= 5:
        return "Lost Customers"

    else:
        return "Others"


rfm['Segment'] = rfm['RFM_Score'].apply(get_segment)

print("Segment created successfully!")

rfm[['RFM_Score', 'Segment']].head()
```

```python
rfm['AverageOrderValue'] = (
    rfm['Monetary'] / rfm['Frequency']
)

rfm['PurchaseFrequency'] = rfm['Frequency']

rfm['CLV'] = (
    rfm['AverageOrderValue'] *
    rfm['PurchaseFrequency']
)

rfm[['Monetary',
     'Frequency',
     'AverageOrderValue',
     'CLV']].head()
```

```python
promotion = {
    "Champions": "20% discount",
    "Loyal Customers": "15% discount",
    "Potential Loyalists": "10% discount",
    "At Risk": "25% discount",
    "Lost Customers": "30% discount",
    "Others": "10% discount"
}

rfm['Promotion'] = rfm['Segment'].map(promotion)

rfm[['Segment', 'Promotion']].head()
```

```python
promotion = {
    "Champions": "20% discount",
    "Loyal Customers": "15% discount",
    "Potential Loyalists": "10% discount",
    "At Risk": "25% discount",
    "Lost Customers": "30% discount",
    "Others": "10% discount"
}

rfm['Promotion'] = rfm['Segment'].map(promotion)

rfm[['Segment', 'Promotion']].head()
```

```python
def generate_email(customer_name, segment, promotion, clv):

    if segment == "Champions":
        message = (
            "Thank you for being one of our most valuable customers! "
            "We truly appreciate your continued support."
        )

    elif segment == "Loyal Customers":
        message = (
            "We appreciate your continued loyalty and would love "
            "to make your next shopping experience even better."
        )

    elif segment == "Potential Loyalists":
        message = (
            "We noticed your recent interest in our products. "
            "We would love to see you shop with us again."
        )

    elif segment == "At Risk":
        message = (
            "We miss seeing you! Come back and enjoy a special "
            "offer created especially for you."
        )

    elif segment == "Lost Customers":
        message = (
            "We haven't seen you for a while and would love to "
            "welcome you back."
        )

    else:
        message = (
            "Thank you for shopping with us. "
            "We have a special offer for your next purchase."
        )

    email = f"""
Subject: A Special Offer Just for You, {customer_name}!

Hi {customer_name},

{message}

As a valued customer, you can enjoy an exclusive
{promotion} on your next purchase.

Your estimated customer lifetime value is ₹{clv:,.2f}.

Don't miss this opportunity!

Best regards,
Customer Success Team
"""

    return email
```

```python
customer_names = df.groupby('CustomerID')['CustomerName'].first()

rfm['CustomerName'] = rfm.index.map(customer_names)

rfm[['CustomerName', 'Segment']].head()
```

```python
customer_names = df.groupby('CustomerID')['CustomerName'].first()

rfm['CustomerName'] = rfm.index.map(customer_names)

rfm[['CustomerName', 'Segment']].head()
```

```python
rfm['PersonalizedEmail'] = rfm.apply(
    lambda row: generate_email(
        row['CustomerName'],
        row['Segment'],
        row['Promotion'],
        row['CLV']
    ),
    axis=1
)

print("Personalized emails generated successfully!")
```
