```python
print("Evaluation of Personalization, Relevance and Consistency")
```

```python
# Evaluation criteria

criteria = [
    "Personalization",
    "Relevance",
    "Consistency"
]

for i, criterion in enumerate(criteria, 1):
    print(i, ".", criterion)
```

```python
# Evaluation scores

personalization_score = 0
relevance_score = 0
consistency_score = 0

print("Personalization Score:", personalization_score)
print("Relevance Score:", relevance_score)
print("Consistency Score:", consistency_score)
```

```python
# Evaluation scale

print("Evaluation Scale")
print("1 = Poor")
print("2 = Average")
print("3 = Good")
print("4 = Very Good")
print("5 = Excellent")
```

```python
# Example evaluation scores

personalization_score = 5
relevance_score = 5
consistency_score = 5

overall_score = (
    personalization_score +
    relevance_score +
    consistency_score
) / 3

print("Personalization:", personalization_score, "/ 5")
print("Relevance:", relevance_score, "/ 5")
print("Consistency:", consistency_score, "/ 5")
print("Overall Score:", overall_score, "/ 5")
```

```python
# Final evaluation result

print("Final Evaluation")
print("-----------------")
print("Personalization :", personalization_score, "/ 5")
print("Relevance       :", relevance_score, "/ 5")
print("Consistency     :", consistency_score, "/ 5")
print("Overall Score   :", round(overall_score, 2), "/ 5")
```

```python
# Evaluation conclusion

if overall_score >= 4:
    result = "The generated emails show good personalization, relevance, and consistency."
elif overall_score >= 3:
    result = "The generated emails show moderate personalization, relevance, and consistency."
else:
    result = "The generated emails need improvement."

print("Conclusion:")
print(result)
```

```python
# Final summary

print("====================================")
print("EMAIL GENERATOR EVALUATION SUMMARY")
print("====================================")
print("Personalization :", personalization_score, "/ 5")
print("Relevance       :", relevance_score, "/ 5")
print("Consistency     :", consistency_score, "/ 5")
print("Overall Score   :", round(overall_score, 2), "/ 5")
print("====================================")
print(result)
```

```python
import matplotlib.pyplot as plt

criteria = ["Personalization", "Relevance", "Consistency"]
scores = [personalization_score, relevance_score, consistency_score]

plt.figure(figsize=(7, 4))
plt.bar(criteria, scores)
plt.ylim(0, 5)
plt.xlabel("Evaluation Criteria")
plt.ylabel("Score")
plt.title("Personalized Email Evaluation")
plt.show()
```

```python
print("Evaluation completed successfully.")
print("The personalized email generator was evaluated based on:")
print("1. Personalization")
print("2. Relevance")
print("3. Consistency")
```

```python
# Save evaluation results

evaluation_result = {
    "Personalization": personalization_score,
    "Relevance": relevance_score,
    "Consistency": consistency_score,
    "Overall_Score": round(overall_score, 2)
}

print(evaluation_result)
```
