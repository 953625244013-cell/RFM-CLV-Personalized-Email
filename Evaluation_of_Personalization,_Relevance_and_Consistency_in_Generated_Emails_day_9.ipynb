```python
emails = [
    {
        "customer": "Priya",
        "segment": "Champions",
        "email": "Hi Priya, enjoy an exclusive 20% discount on your next purchase. Thank you for being one of our valued customers!"
    },
    {
        "customer": "Arun",
        "segment": "Loyal Customers",
        "email": "Hi Arun, we appreciate your loyalty. Enjoy 15% off on your next purchase!"
    },
    {
        "customer": "Kavi",
        "segment": "At Risk",
        "email": "Hi Kavi, we miss you! Come back and enjoy 25% off on your next purchase."
    }
]

print("Sample emails created successfully.")
```

```python
offer_map = {
    "Champions": "20%",
    "Loyal Customers": "15%",
    "At Risk": "25%"
}

def evaluate_relevance(data):
    results = []

    for item in data:
        expected_offer = offer_map.get(item["segment"], "")
        relevant = expected_offer in item["email"]

        results.append({
            "Customer": item["customer"],
            "Segment": item["segment"],
            "Relevant": "Yes" if relevant else "No"
        })

    return results

relevance_results = evaluate_relevance(emails)

for result in relevance_results:
    print(result)
```

```python
offer_map = {
    "Champions": "20%",
    "Loyal Customers": "15%",
    "At Risk": "25%"
}

def evaluate_relevance(data):
    results = []

    for item in data:
        expected_offer = offer_map.get(item["segment"], "")
        relevant = expected_offer in item["email"]

        results.append({
            "Customer": item["customer"],
            "Segment": item["segment"],
            "Relevant": "Yes" if relevant else "No"
        })

    return results

relevance_results = evaluate_relevance(emails)

for result in relevance_results:
    print(result)
```

```python
def evaluate_consistency(data):
    results = []

    for item in data:
        has_greeting = item["email"].startswith("Hi")
        has_offer = "%" in item["email"]
        has_customer_name = item["customer"].lower() in item["email"].lower()

        consistent = has_greeting and has_offer and has_customer_name

        results.append({
            "Customer": item["customer"],
            "Consistent": "Yes" if consistent else "No"
        })

    return results

consistency_results = evaluate_consistency(emails)

for result in consistency_results:
    print(result)
```

```python
personalization_score = 0
relevance_score = 0
consistency_score = 0

for result in personalization_results:
    score = int(result["Personalization Score"].split("/")[0])
    personalization_score += score / 2

for result in relevance_results:
    if result["Relevant"] == "Yes":
        relevance_score += 1

for result in consistency_results:
    if result["Consistent"] == "Yes":
        consistency_score += 1

n = len(emails)

personalization_percentage = (personalization_score / n) * 100
relevance_percentage = (relevance_score / n) * 100
consistency_percentage = (consistency_score / n) * 100

overall_score = (
    personalization_percentage +
    relevance_percentage +
    consistency_percentage
) / 3

print("Personalization:", round(personalization_percentage, 2), "%")
print("Relevance:", round(relevance_percentage, 2), "%")
print("Consistency:", round(consistency_percentage, 2), "%")
print("Overall Evaluation Score:", round(overall_score, 2), "%")
```

```python
import pandas as pd

evaluation_data = []

for i in range(len(emails)):
    evaluation_data.append({
        "Customer": emails[i]["customer"],
        "Segment": emails[i]["segment"],
        "Personalization": personalization_results[i]["Personalization Score"],
        "Relevance": relevance_results[i]["Relevant"],
        "Consistency": consistency_results[i]["Consistent"]
    })

evaluation_df = pd.DataFrame(evaluation_data)

display(evaluation_df)
```

```python
import matplotlib.pyplot as plt

criteria = ["Personalization", "Relevance", "Consistency"]
scores = [
    personalization_percentage,
    relevance_percentage,
    consistency_percentage
]

plt.figure(figsize=(8, 5))
plt.bar(criteria, scores)
plt.ylim(0, 100)
plt.xlabel("Evaluation Criteria")
plt.ylabel("Score (%)")
plt.title("Email Quality Evaluation")
plt.show()
```
