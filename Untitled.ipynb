```python
# Product / Service Information

product_info = {
    "Product": "Personalized Customer Offers",
    "Service": "Customer Retention and Promotional Email Service",
    "Purpose": "Provide personalized offers based on customer segments",
    "Benefits": [
        "Improves customer engagement",
        "Supports customer retention",
        "Provides personalized offers"
    ]
}

# Communication Guidelines

communication_guidelines = {
    "Tone": "Friendly, professional and personalized",
    "Language": "Simple and clear",
    "Content": "Mention customer value and relevant offer",
    "Avoid": "Generic and unnecessary information",
    "Call_to_Action": "Encourage the customer to use the offer"
}

# Segment-based Offers

segment_offers = {
    "Champions": "20% OFF",
    "Loyal Customers": "15% OFF",
    "Potential Loyalists": "10% OFF",
    "At Risk": "25% OFF",
    "Lost Customers": "30% OFF",
    "Others": "10% OFF"
}

print("Product Information:", product_info)
print("\nCommunication Guidelines:", communication_guidelines)
print("\nSegment Offers:", segment_offers)
```

```python
# Personalized Email Generator

def get_offer(segment):
    return segment_offers.get(segment, "10% OFF")

def create_email(customer_name, segment):
    offer = get_offer(segment)

    return f"""
Hi {customer_name},

Thank you for being a valued customer!

As a {segment}, we have a special offer for you:
{offer} on your next purchase.

Don't miss this opportunity. We look forward to serving you again!

Best Regards,
Customer Success Team
"""

# Example
customer_name = "Dharshini"
customer_segment = "Loyal Customers"

email = create_email(customer_name, customer_segment)

print(email)
```

```python
# Customer data

customers = [
    {"Name": "Arun", "Segment": "Champions"},
    {"Name": "Priya", "Segment": "Loyal Customers"},
    {"Name": "Kavi", "Segment": "Potential Loyalists"},
    {"Name": "Dharshini", "Segment": "At Risk"},
    {"Name": "Nithya", "Segment": "Lost Customers"}
]

customers
```
