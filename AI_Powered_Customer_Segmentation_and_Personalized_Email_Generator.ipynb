```python
# Segment-specific email prompts

segment_prompts = {
    "Champions":
        "Thank the loyal customer and provide an exclusive reward.",

    "Loyal Customers":
        "Appreciate their continued purchases and encourage repeat purchases.",

    "Potential Loyalists":
        "Encourage the customer to become a regular customer with an attractive offer.",

    "At Risk":
        "Encourage the customer to return by highlighting a special offer.",

    "Lost Customers":
        "Try to win back the customer with a strong promotional offer.",

    "Others":
        "Create a simple promotional message to increase customer engagement."
}

print(segment_prompts)
```

```python
# Create prompt for each customer segment

def create_segment_prompt(customer_name, segment, offer):

    guideline = segment_prompts.get(
        segment,
        "Create a friendly promotional message."
    )

    return f"""
Create a short personalized marketing email.

Customer: {customer_name}
Segment: {segment}
Offer: {offer}

Communication guideline:
{guideline}

Requirements:
- Use a friendly and professional tone.
- Keep the email short and clear.
- Mention the offer.
- Include a call-to-action.
- End with "Best Regards, Customer Success Team".
"""
```

```python
# Generate prompts for all customer segments

segment_prompts = {
    "Champions": "Thank the loyal customer and provide an exclusive reward.",
    "Loyal Customers": "Appreciate their continued purchases and encourage repeat purchases.",
    "Potential Loyalists": "Encourage the customer to become a regular customer.",
    "At Risk": "Encourage the customer to return with a special offer.",
    "Lost Customers": "Try to win back the customer with a promotional offer.",
    "Others": "Create a simple promotional message to increase engagement."
}

segment_offers = {
    "Champions": "20% OFF",
    "Loyal Customers": "15% OFF",
    "Potential Loyalists": "10% OFF",
    "At Risk": "25% OFF",
    "Lost Customers": "30% OFF",
    "Others": "10% OFF"
}

for segment in segment_prompts:
    offer = segment_offers[segment]
    guideline = segment_prompts[segment]

    prompt = f"""
Create a short personalized marketing email.

Customer: Customer
Segment: {segment}
Offer: {offer}

Guideline:
{guideline}

Use a friendly and professional tone.
Mention the offer clearly.
Include a call-to-action.
End with:
Best Regards,
Customer Success Team
"""

    print("=" * 60)
    print("SEGMENT:", segment)
    print(prompt)
```

```python
# Generate personalized emails using segment-specific prompts

import pandas as pd

customers = [
    {"Name": "Arun", "Segment": "Champions"},
    {"Name": "Priya", "Segment": "Loyal Customers"},
    {"Name": "Kavi", "Segment": "Potential Loyalists"},
    {"Name": "Dharshini", "Segment": "At Risk"},
    {"Name": "Nithya", "Segment": "Lost Customers"}
]

segment_offers = {
    "Champions": "20% OFF",
    "Loyal Customers": "15% OFF",
    "Potential Loyalists": "10% OFF",
    "At Risk": "25% OFF",
    "Lost Customers": "30% OFF",
    "Others": "10% OFF"
}

segment_prompts = {
    "Champions": "Thank the loyal customer and provide an exclusive reward.",
    "Loyal Customers": "Appreciate their continued purchases and encourage repeat purchases.",
    "Potential Loyalists": "Encourage the customer to become a regular customer.",
    "At Risk": "Encourage the customer to return with a special offer.",
    "Lost Customers": "Try to win back the customer with a promotional offer."
}

segment_emails = []

for customer in customers:

    name = customer["Name"]
    segment = customer["Segment"]

    offer = segment_offers.get(segment, "10% OFF")

    guideline = segment_prompts.get(
        segment,
        "Create a friendly promotional message."
    )

    email = f"""
Hi {name},

Thank you for being a valued customer!

As a {segment}, we have a special offer for you:
{offer}

{guideline}

Don't miss this special opportunity. We look forward to serving you again!

Best Regards,
Customer Success Team
"""

    segment_emails.append({
        "Customer": name,
        "Segment": segment,
        "Offer": offer,
        "Email": email
    })

segment_email_df = pd.DataFrame(segment_emails)

display(segment_email_df)
```

```python
# Preview segment-specific emails

for _, row in segment_email_df.iterrows():
    print("=" * 60)
    print("Customer :", row["Customer"])
    print("Segment  :", row["Segment"])
    print("Offer    :", row["Offer"])
    print("\nEmail:")
    print(row["Email"])
```

```python
# Validate segment-specific emails

def check_email(email):
    return (
        len(str(email)) > 20
        and "offer" in str(email).lower()
        and "best regards" in str(email).lower()
    )

segment_email_df["Valid"] = segment_email_df["Email"].apply(check_email)

print("Email Validation Results:")
display(segment_email_df[[
    "Customer",
    "Segment",
    "Offer",
    "Valid"
]])
```

```python
# Segment-wise email summary

summary = (
    segment_email_df
    .groupby("Segment")
    .agg(
        Customers=("Customer", "count"),
        Valid_Emails=("Valid", "sum")
    )
    .reset_index()
)

display(summary)
```

```python
# Final segment-specific prompt report

for segment in segment_prompts:
    print("=" * 50)
    print("SEGMENT :", segment)
    print("GUIDELINE:", segment_prompts[segment])
    print("OFFER    :", segment_offers.get(segment, "10% OFF"))

print("\nSegment-specific prompt design completed successfully!")
```

```python
# Final segment-specific email report

final_segment_report = segment_email_df[
    ["Customer", "Segment", "Offer", "Email", "Valid"]
]

display(final_segment_report)

print("Final segment-specific email report generated successfully!")
```

```python
# Save final segment-specific email report

final_segment_report.to_csv(
    "Final_Segment_Specific_Email_Report.csv",
    index=False
)

print("Final segment-specific email report saved successfully!")
```
