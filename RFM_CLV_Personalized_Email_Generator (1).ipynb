# Personalized Email Workflow

The personalized email generation process follows these steps:

1. **Customer Data Collection**
   - Customer purchase information is loaded into the system.

2. **RFM Analysis**
   - Recency, Frequency, and Monetary values are calculated.

3. **Customer Segmentation**
   - Customers are classified into segments such as Champions, Loyal Customers, Potential Loyalists, At Risk, and Lost Customers.

4. **CLV Calculation**
   - Customer Lifetime Value is calculated to estimate the customer's value.

5. **Promotion Selection**
   - A suitable promotion is assigned based on the customer segment.

6. **LLM Input**
   - Customer name, segment, CLV, promotion, and segment-specific strategy are provided to the LLM.

7. **Personalized Email Generation**
   - The LLM generates a customized email subject and email body.

8. **Structured Output**
   - The LLM response is returned in JSON format.

9. **Final Output**
   - Customer details, segment, CLV, promotion, email subject, and email body are stored in the final dataset.
